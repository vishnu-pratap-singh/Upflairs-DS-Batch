* MAE
* MSE
* RMSE
* R2 SCORE 


#### MEAN ABSOLUTE ERROR : 
* THE AVERAGE OF ABSOLUTE DIFFERENCE BETWEEN ACTUAL AND PREDICTED OUTCOMES
#### MEAN SQUARE ERROR:
* THE AVERAGE OF SQUARE DIFFERENCE OF ACTUAL AND PREDICTED(I/N*(ACTUAL-REDICTED)2
####  ROOT MEAN SQUARE ERROR:
* ROOT OF MSE
#### r2 SCORE:
* COEFFICIENT OF DETERMINATION R² = 1- ((ACTUAL-PREDICTED)/(ACTUAL-MEAN))

# Question 4 Linear Regression model 


```python
import pandas as pd

```

```python
data={
    "Hour": [1,2,3,4,5],
    "Marks": [30,40,50,60,70]
}

df= pd.DataFrame(data)
df.head()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
# create a instance of LinearRegression 
X = df[['Hour']]  
Y= df['Marks']
model = LinearRegression()
model.fit(X, Y)

```

```python
prediction =model.predict(X)
print(prediction)
```

```python
import numpy as np

```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
```

```python
mae= mean_absolute_error(Y, prediction)
print(mae)

mse = mean_squared_error(Y, prediction)
print(mse)

rmse = np.sqrt(mse)
print(rmse)

from sklearn.metrics import r2_score
r2 = r2_score(Y, prediction)
print(r2)
```
